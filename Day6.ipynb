```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned dataset created in Day 5
df = pd.read_csv("student_scores_cleaned.csv")

print("Dataset loaded successfully for visualization!")
print(df.head())
```

```python
# Scatter plot to show relationship between Study Hours and Scores
plt.figure(figsize=(8, 5))
plt.scatter(df['StudyHours'], df['Score'], color='blue', marker='o', s=70)

plt.title('Study Hours vs Percentage Score', fontsize=14)
plt.xlabel('Hours Studied', fontsize=12)
plt.ylabel('Percentage Score', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
# Create subplots for Line Plot and Bar Plot side-by-side
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Line Plot (Score Trend across Students)
axes[0].plot(df['Student'], df['Score'], marker='o', color='green', linewidth=2)
axes[0].set_title('Score Trend per Student')
axes[0].set_xlabel('Student Name')
axes[0].set_ylabel('Score')
axes[0].tick_params(axis='x', rotation=45)

# 2. Bar Plot (Study Hours per Student)
axes[1].bar(df['Student'], df['StudyHours'], color='orange', edgecolor='black')
axes[1].set_title('Study Hours per Student')
axes[1].set_xlabel('Student Name')
axes[1].set_ylabel('Hours Studied')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap to check feature relationships
plt.figure(figsize=(6, 4))
sns.heatmap(df[['StudyHours', 'Score']].corr(), annot=True, cmap='Blues', fmt=".2f")

plt.title('Correlation Heatmap')
plt.show()
```

```python
# Save all charts

# Scatter Plot
plt.figure(figsize=(8,5))
plt.scatter(df["StudyHours"], df["Score"], color="blue")
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title("Study Hours vs Score")
plt.grid(True)
plt.savefig("scatter_plot.png")
plt.close()

# Line Chart
plt.figure(figsize=(8,5))
plt.plot(df["Student"], df["Score"], marker="o", color="green")
plt.xticks(rotation=45)
plt.title("Score Trend")
plt.savefig("line_chart.png")
plt.close()

# Bar Chart
plt.figure(figsize=(8,5))
plt.bar(df["Student"], df["StudyHours"], color="orange")
plt.xticks(rotation=45)
plt.title("Study Hours")
plt.savefig("bar_chart.png")
plt.close()

print("All charts saved successfully!")
```
