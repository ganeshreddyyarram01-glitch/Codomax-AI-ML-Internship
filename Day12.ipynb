### Project Description
This notebook implements an end-to-end Machine Learning pipeline using **Simple Linear Regression** to model the relationship between daily study hours and student exam performance.

#### Workflow Overview:
1. Data Loading & Exploratory Data Analysis (EDA)
2. Model Training & Validation Split
3. Statistical Performance Evaluation
4. Deployment-ready Prediction Function

```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# Global Configuration Constants
DATA_PATH = "student_scores_cleaned.csv"
RANDOM_STATE = 42
TEST_SIZE = 0.2

# Visual Styling Setup
sns.set_theme(style="whitegrid")
print("Environment configured and dependencies successfully loaded.")
```

```python
def load_and_preprocess(filepath: str):
    """Loads dataset and extracts feature matrix (X) and target vector (y)."""
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"Dataset file standard path missing: {filepath}")

    df = pd.read_csv(filepath)
    X = df[["StudyHours"]]
    y = df["Score"]
    return df, X, y


def train_regression_model(X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE):
    """Splits dataset into train/test sets and fits a Linear Regression model."""
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state
    )

    model = LinearRegression()
    model.fit(X_train, y_train)

    return model, X_train, X_test, y_train, y_test


# Execute Pipeline Stage 1 & 2
df, X, y = load_and_preprocess(DATA_PATH)
model, X_train, X_test, y_train, y_test = train_regression_model(X, y)
print("Data loaded and Linear Regression model trained successfully.")
```

```python
def evaluate_model(model, X_test, y_test):
    """Computes standard evaluation metrics and logs summary output."""
    y_pred = model.predict(X_test)

    metrics = {
        "MAE": mean_absolute_error(y_test, y_pred),
        "MSE": mean_squared_error(y_test, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, y_pred)),
        "R2 Score": r2_score(y_test, y_pred),
    }

    print("------------------------------------------")
    print("        MODEL EVALUATION SUMMARY          ")
    print("------------------------------------------")
    for metric, value in metrics.items():
        print(f"{metric:<10}: {value:.4f}")
    print("------------------------------------------")

    return y_pred, metrics


y_pred, evaluation_results = evaluate_model(model, X_test, y_test)
```

```python
def predict_score(hours: float) -> float:
    """Predicts percentage score based on daily study hours input."""
    if not (0 <= hours <= 24):
        raise ValueError("Input hours must be between 0 and 24.")

    input_data = pd.DataFrame({"StudyHours": [hours]})
    raw_pred = model.predict(input_data)[0]

    # Constrain output within realistic limits (0% to 100%)
    return float(np.clip(raw_pred, 0, 100))


# Demonstration of deployment helper function
example_hours = 7.5
print(f"Predicted Score for {example_hours} hrs: {predict_score(example_hours):.2f}%")
```
