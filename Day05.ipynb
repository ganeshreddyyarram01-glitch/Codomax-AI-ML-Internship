```python
import pandas as pd

# Load the dataset
df = pd.read_csv("student_scores.csv")

print("--- Initial Dataset Snapshot ---")
print(df.head(10))
print("\nTotal rows before cleaning:", len(df))
```

```python
# 1. Check for missing values in each column
print("--- Missing Values Count ---")
print(df.isnull().sum())

# 2. Handle missing values (if any exist)
# Option A: Drop missing rows (recommended if very few)
df_clean = df.dropna()

# Option B: Impute missing numerical values with mean (alternative)
# df['StudyHours'] = df['StudyHours'].fillna(df['StudyHours'].mean())

print("\n--- Cleaned Missing Values Snapshot ---")
print(df_clean.isnull().sum())
```

```python
# 1. Identify duplicate rows
duplicates_count = df_clean.duplicated().sum()
print(f"Number of duplicate rows found: {duplicates_count}")

# 2. Drop duplicate rows (keeping first occurrence)
df_clean = df_clean.drop_duplicates()

print(f"Total rows after removing duplicates: {len(df_clean)}")
```

```python
# Detailed dataset statistics for numerical features
print("--- Dataset Descriptive Statistics ---")
print(df_clean.describe())

# Save the cleaned dataset for upcoming Machine Learning days
df_clean.to_csv("student_scores_cleaned.csv", index=False)
print("\n✅ Cleaned dataset successfully saved to 'student_scores_cleaned.csv'!")
```
